```python
import pandas as pd

df = pd.read_csv("../data/raw/bank_churn.csv")
```

```python
df["churn"].value_counts(normalize=True)
```

```python
df = df.drop(columns="customer_id")
```

```python
print(df["country"].isna().sum())
print(df["country"].unique())
print(df["country"].value_counts())
print(df["country"].describe())
print(df.groupby("country")["churn"].mean())
# categorical -> OneHotEncoder
```

```python
print(df["gender"].isna().sum())
print(df["gender"].unique())
print(df["gender"].value_counts())
print(df["gender"].describe())
print(df.groupby("gender")["churn"].mean())
```

```python
df["gender"] = df["gender"].map({"Female": 0, "Male": 1})
df["gender"].unique()
```

```python
print(df["age"].isna().sum())
print(df["age"].describe())
print(df[["age", "churn"]].corr())
print(df.groupby(pd.cut(df["age"], bins=[-1, 18, 30, 40, 50, 100]))["churn"].mean())
# StandardScaler
```

```python
print(df["tenure"].isna().sum())
print(df["tenure"].unique())
print(df["tenure"].value_counts())
print(df[["tenure", "churn"]].corr())
print(df.groupby("tenure")["churn"].mean())
# no signal
df = df.drop(columns="tenure")
```

```python
print(df["balance"].isna().sum())
print(df["balance"].describe())
print(df[["balance", "churn"]].corr())
print(df.groupby(pd.cut(df["balance"], bins=[-1, 0, 50000, 100000, 200000, 300000]))["churn"].mean())
# StandardScaler
```

```python
print(df["products_number"].isna().sum())
print(df["products_number"].describe())
print(df[["products_number", "churn"]].corr())
print(df.groupby("products_number")["churn"].mean())
print(df["products_number"].value_counts())
```

```python
df["is_many_products"] = df["products_number"].map({1: 0, 2: 0, 3: 1, 4: 1})
print(df[["is_many_products", "churn"]].corr())
print(df.groupby("is_many_products")["churn"].mean())
```

```python
print(df["credit_card"].isna().sum())

print(df.groupby("credit_card")["churn"].mean())
print(df[["credit_card", "churn"]].corr())
# no signal
df = df.drop(columns="credit_card")
```

```python
print(df["estimated_salary"].isna().sum())
print(df["estimated_salary"].describe())
print(df[["estimated_salary", "churn"]].corr())
print(df.groupby(pd.cut(df["estimated_salary"], bins=[10, 100, 1000, 10000, 100000, 200000]))["churn"].mean())
# no signal
df = df.drop(columns="estimated_salary")
```

```python
print(df["active_member"].isna().sum())
print(df[["active_member", "churn"]].corr())
print(df.groupby("active_member")["churn"].mean())
```
